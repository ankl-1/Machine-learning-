```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import SGDRegressor
from sklearn.preprocessing import StandardScaler
x_train=np.array([[1],[2],[3],[4],[5]])
y_train=np.array([120,220,300,450,500])
scaler=StandardScaler()
x_norm=scaler.fit_transform(x_train)
model=SGDRegressor(max_iter=1000,random_state=42)
model.fit(x_norm,y_train)
new_x=scaler.transform(np.array([[6],[7]]))
predict=model.predict(new_x)
print("weight:",model.coef_)
print("bias:",model.intercept_)
for x, pred in zip([6,7],predict):
    print(f"Input: {x} | Predicted output: {pred:.2f}")
x_line=np.linspace(1,7,100).reshape(-1,1)
x_line_norm=scaler.transform(x_line)
y_line=model.predict(x_line_norm)
plt.scatter(x_train,y_train,label="Training data")
plt.plot(x_line,y_line,label="Regression line")
plt.scatter([6,7], predict, marker="x",s=100,label="New predictions")
plt.xlabel("Input")
plt.ylabel("Output")
plt.title("Linear Regression using Scikit-learn")
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sklearn.metrics import mean_squared_error, r2_score
train_predictions=model.predict(x_norm)
mse = mean_squared_error(y_train, train_predictions)
r2 = r2_score(y_train, train_predictions)
print(f"Training MSE: {mse:.2f}")
print(f"Training R² Score: {r2:.4f}")
```
