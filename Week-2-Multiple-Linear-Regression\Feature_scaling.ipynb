```python
import numpy as np
import matplotlib.pyplot as plt
x_train = np.array([[2104, 5, 1, 45],[1416, 3, 2, 40],[852,  2, 1, 35]], dtype=float)
y_train = np.array([460, 232, 178], dtype=float)
print("Original features:")
print(x_train)
print("\nShape of x_train:", x_train.shape)
mean=np.mean(x_train,axis=0)
std_dev=np.std(x_train,axis=0)
print(f"mean:{mean}\nStandard diviation:{std_dev}")
def z_norm(X):
  x_mean=np.mean(X,axis=0)
  x_std=np.std(X,axis=0)
  x_norm=(X-x_mean)/x_std
  return x_norm,x_mean,x_std
x_norm,x_mean,x_std=z_norm(x_train)
print(f"Original features are:\n{x_train}")
print(f"Normalised features are:\n{x_norm}")
print(f"Mean of normalised features are:\n{x_mean}")
print(f"Standard deviation of normalised features are:\n")
print(np.mean(x_norm,axis=0))
print(f"Standard deviation of normalised features are:\n")
print(np.std(x_norm,axis=0))
feature_names = ["Size", "Bedrooms", "Floors", "Age"]
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for j in range(4):
    axes[0].plot(x_train[:, j],marker="o",label=feature_names[j])
    axes[1].plot(x_norm[:, j],marker="o",label=feature_names[j])
axes[0].set_title("Original Features")
axes[0].set_xlabel("House Index")
axes[0].set_ylabel("Feature Value")
axes[0].legend()
axes[0].grid(True)
axes[1].set_title("After Z-Score Normalization")
axes[1].set_xlabel("House Index")
axes[1].set_ylabel("Normalized Value")
axes[1].legend()
axes[1].grid(True)
plt.tight_layout()
plt.show()
```
