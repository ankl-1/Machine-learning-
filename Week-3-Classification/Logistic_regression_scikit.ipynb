```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression

X = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y = np.array([0, 0, 0, 1, 1, 1])

lr_model=LogisticRegression()
lr_model.fit(X,y)
y_pred=lr_model.predict(X)

print(f"Actual values: {y}")
print(f"\nPredicted values: {y_pred}")
print(f"\nAccuracy of the model :{lr_model.score(X,y)}\n")

print(f"Model coefficirnt :{lr_model.coef_},\nModel Intercept : {lr_model.intercept_}\n")
print(lr_model.predict_proba(X))
```
