```python
import numpy as np
import matplotlib.pyplot as plt
input_arr=np.array([1,2,3])
exp_arr=np.array(np.exp(input_arr))
print("Input array",input_arr)
print("Exponent of the array:",exp_arr)
def sigmoid(z):
  g=1/(1+(np.exp(-z)))
  return g
z_temp=np.arange(-10,10)
y=sigmoid(z_temp)
print("\nSigmoid functions:\n",y)

fig,ax=plt.subplots(1,1,figsize=(5,3))
ax.plot(z_temp,y,color="blue")
ax.set_title("Sigmoid function")
ax.set_xlabel("z")
ax.set_ylabel("g(z)")
plt.grid(True)
plt.show()
```
