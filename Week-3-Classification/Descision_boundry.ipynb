```python
import numpy as np
import matplotlib.pyplot as plt
X = np.array([[0.5,1.5],[1,1],[1.5,0.5],[3,0.5],[2,2],[1,2.5]])
y = np.array([0,0,0,1,1,1]).reshape(-1,1)

fig,ax=plt.subplots(1,1,figsize=(6,6))
positive=y.flatten()==1
negative=y.flatten()==0
ax.scatter(X[positive,0],X[positive,1],marker='o',c='b')
ax.scatter(X[negative,0],X[negative,1],marker='x',c='r')
ax.axis([0,4,0,3.5])
ax.set_ylabel('$x_1$')
ax.set_xlabel('$x_0$')
plt.show()


```

```python
def sigmoid(z):
  g=1/(1+(np.exp(-z)))
  return g
z=np.arange(-10,10)
fig,ax=plt.subplots(1,1,figsize=(6,6))
ax.plot(z,sigmoid(z),c="b")
ax.set_title("Sigmoid Function")
ax.set_xlabel("z")
ax.set_ylabel("Sigmoid(z)")
ax.axvline(x=0, color="gray", linestyle="--")
plt.show()
```

```python
x0=np.arange(0,6)
x1=3 -x0
fig,ax=plt.subplots(1,1,figsize=(6,6))
ax.plot(x0,x1,color="blue",label="Decision boundary")
ax.axis([0,4,0,3.5])
ax.fill_between(x0,x1,alpha=0.2)
positive=y.flatten()==1
negative=y.flatten()==0
ax.scatter(X[negative,0],X[negative,1],marker="x",color="red",label="y = 0")
ax.scatter(X[positive,0],X[positive,1],marker="o",color="green",label="y = 1")
ax.set_ylabel(r"$x_1$")
ax.set_xlabel(r"$x_0$")
ax.legend()
plt.show()
```
