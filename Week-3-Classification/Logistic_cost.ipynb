```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y = np.array([0,0,0,1,1,1])

fig,ax=plt.subplots(1,1,figsize=(6,6))
positive=y.flatten()==1
negative=y.flatten()==0
plt.scatter(X[positive,0],X[positive,1],marker='o',c='b',label="y=1")
plt.scatter(X[negative,0],X[negative,1],marker='x',c='r',label="y=0")
ax.axis([0,4,0,3.5])
ax.set_ylabel('$x_1$',fontsize=12)
ax.set_xlabel('$x_0$',fontsize=12)
ax.grid(True)
ax.legend()
plt.show()


```

```python
def sigmoid(z):
  g=1/(1+np.exp(-z))
  return g

def compute_logistic_cost(x,y,w,b):
  cost=0.0
  m=x.shape[0]
  for i in range(m):
    z=np.dot(x[i],w)+b
    f_wb=sigmoid(z)
    cost+=-y[i]*(np.log(f_wb))-(1-y[i])*(np.log(1-f_wb))
  cost/=m
  return cost

```

```python
w=np.array([1.0,1.0])
b=-3.0

cost=compute_logistic_cost(X,y,w,b)
print(f"Cost:{cost:.4f}")
```

```python
#Effect of change in bias while keeping the weights same:

x0=np.arange(0,6)
x1=3-x0
x1_other=4-x0
fig, ax=plt.subplots(1,1,figsize=(6,6))
ax.plot(x0,x1,color="blue",label="$b=-3$")
ax.plot(x0,x1_other,color="magenta",label="$b=-4$")
ax.axis([0, 4, 0, 4])
positive=y.flatten()==1
negative=y.flatten()==0
ax.scatter(X[positive,0],X[positive,1],marker="o",color="blue",label="y=1")
ax.scatter(X[negative,0],X[negative,1],marker="x",color="red",label="y=0")
ax.set_ylabel("$x_1$",fontsize=12)
ax.set_xlabel("$x_0$",fontsize=12)
plt.grid(True)
ax.legend(loc="upper right")
ax.set_title("Decision Boundary")
plt.show()
```

```python
w_array1 = np.array([1,1])
b_1=-3
w_array2 = np.array([1,1])
b_2=-4

print("Cost for b=-3 : ", compute_logistic_cost(X,y,w_array1,b_1))
print("Cost for b=-4 : ", compute_logistic_cost(X,y,w_array2,b_2))
```
